#Q2)  In a class of 50 students:
20 like Mathematics (M)
15 like Science (S)
5 like both subjects
#What is the probability that a student chosen at random likes Mathematics or Science
#Ans-->
Let’s solve this step by step 📘  

We are given:  
- Total students = 50  
- Students who like Mathematics (M) = 20  
- Students who like Science (S) = 15  
- Students who like both = 5  

Total students = 50
Mathematics (M) = 20
Science (S) = 15
Both (M ∩ S) = 5

n(M ∪ S) = n(M) + n(S) - n(M ∩ S)
          = 20 + 15 - 5
          = 30

P(M ∪ S) = n(M ∪ S) / Total students
          = 30 / 50
          = 3/5

#Q3) A bag has 3 red and 2 blue balls. If one ball is drawn randomly and is red, what is the probability that the next ball is also red (without replacement)?
#Ans-->


**Step 1: Initial setup**  
- Total balls = 3 red + 2 blue = 5  
- One ball is drawn and it is red.  

**Step 2: Remaining balls after first draw**  
- Red balls left = 3 − 1 = 2  
- Blue balls left = 2  
- Total balls left = 4  

**Step 3: Probability of drawing another red ball**  
Total balls = 3 red + 2 blue = 5
One ball drawn is red → Remaining balls = 4
Remaining red balls = 2

P(Next ball is red) = Remaining red balls / Remaining total balls
                    = 2 / 4
                    = 1/2

#Q4)  The population of a school is divided into 60% boys and 40% girls. If you want equal representation of both genders in the sample, which method should you use: Simple Random Sampling or Stratified Sampling?
Why?
#Ans-->
We are given:
- School population: 60% boys, 40% girls
- Goal: Equal representation of both genders in the sample

Method:
- Simple Random Sampling: Each student has equal chance of selection. This will reflect the actual proportion (≈60% boys, 40% girls), not equal representation.
- Stratified Sampling: The population is divided into strata (boys and girls). Then we select equal numbers from each stratum to ensure balanced representation.

Answer:
Use Stratified Sampling because it allows us to control the proportion of boys and girls in the sample, ensuring equal representation of both genders.

#Q5)  The average height of 1000 students = 160 cm. A sample of 100 students shows an average height = 158 cm.
Find the sampling error
#Ans-->
Population mean (μ) = 160 cm
Sample mean (x̄) = 158 cm

Sampling Error = Sample mean − Population mean
               = 158 − 160
               = −2 cm

Answer: The sampling error is −2 cm (i.e., the sample average is 2 cm less than the population average).


#Q6)The population mean salary is ₹50,000 with σ = ₹5,000. If we take a sample of 100 employees, what is the standard error of the mean (SEM)?
#Ans-->


---

Population Standard Deviation (sigma) = 5000
Sample Size (n) = 100

Standard Error of Mean (SEM) का formula:
SEM = sigma / sqrt(n)

Calculation:
SEM = 5000 / sqrt(100)
SEM = 5000 / 10
SEM = 500

Final Answer:
Standard Error of the Mean (SEM) = 500

#Q7) In a group of 100 students:40 like Cricket (C)30 like Football (F)10 like both Cricket and Football Find the probability that a student likes at least one sport.

Total students = 100

Students who like Cricket (C) = 40
Students who like Football (F) = 30
Students who like both (C ∩ F) = 10

Calculation:

n(C ∪ F) = 60

अब probability निकालते हैं:

Probability = Number who like at least one sport / Total students

Probability = 60 / 100
Probability = 0.6

---

Final Answer:
Probability that a student likes at least one sport = 0.6
(या 60%)

= 4 / 52

#Q9) A factory produces bulbs with 2% defective rate. If 5 bulbs are chosen at random, what is the probability that all are non-defective?
#Ans-->
Defective rate = 2%
So, probability that a bulb is defective = 0.02

Probability that a bulb is non-defective =
= 1 − 0.02
= 0.98

5 bulbs are chosen at random (independent events).

Probability that all 5 bulbs are non-defective:

Probability = (0.98) × (0.98) × (0.98) × (0.98) × (0.98)
Probability = (0.98)^5

Calculate:

(0.98)^5 ≈ 0.9039

---

Final Answer:
Probability that all 5 bulbs are non-defective ≈ 0.9039
(लगभग 90.39%)

#Q10)  Differentiate between discrete and continuous random variables with examples.
#Ans-->

DISCRETE RANDOM VARIABLE:

Definition:
A discrete random variable takes **countable values**.

Nature of values:
Values are finite or countably infinite.

How values occur:
Values come by **counting**.

Probability distribution:
Uses **Probability Mass Function (PMF)**.

Examples:

1. Number of students in a class
2. Number of heads when tossing 3 coins
3. Number of defective items in a batch
4. Number of calls received in one hour

Example values:
0, 1, 2, 3, ...

---

CONTINUOUS RANDOM VARIABLE:

Definition:
A continuous random variable takes **uncountable values** within a range.

Nature of values:
Values lie in an interval and can take **any real number**.

How values occur:
Values come by **measurement**.

Probability distribution:
Uses **Probability Density Function (PDF)**.

Examples:

1. Height of students
2. Weight of a person
3. Time taken to complete a task
4. Temperature of a city

Example values:
50.2, 50.25, 50.251, ...

---

KEY DIFFERENCES (SHORT):

Discrete → Counting → PMF → Exact values
Continuous → Measuring → PDF → Range of values
